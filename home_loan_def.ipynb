```python
print("Primary Key in application_train : SK_ID_CURR")

print("bureau :", bureau.columns[:5])

print("previous_application :", previous_application.columns[:5])

print("pos_CASH_balance :", pos_CASH_balance.columns[:5])

print("credit_card_balance :", credit_card_balance.columns[:5])

print("installments_payments :", installments_payments.columns[:5])
```

```python
bureau.head()
```

```python
bureau.isnull().sum().sort_values(ascending=False).head(20)
```

```python
bureau_agg = bureau.groupby("SK_ID_CURR").agg({

    "SK_ID_BUREAU":"count",

    "DAYS_CREDIT":"mean",

    "AMT_CREDIT_SUM":"mean",

    "AMT_CREDIT_SUM_DEBT":"mean",

    "AMT_CREDIT_SUM_OVERDUE":"mean",

    "CREDIT_DAY_OVERDUE":"mean"

})

bureau_agg.columns=[

"BUREAU_LOAN_COUNT",

"AVG_DAYS_CREDIT",

"AVG_CREDIT_SUM",

"AVG_DEBT",

"AVG_OVERDUE",

"AVG_DAY_OVERDUE"

]

bureau_agg.reset_index(inplace=True)

bureau_agg.head()
```

```python
bureau_balance.head()
```

```python
# merge bureau balance with bureau

bureau_balance_merge = bureau[["SK_ID_BUREAU","SK_ID_CURR"]].merge(

bureau_balance,

on="SK_ID_BUREAU",

how="left"

)

bureau_balance_merge.head()
```

```python
bureau_balance_agg = bureau_balance_merge.groupby("SK_ID_CURR").agg({

"MONTHS_BALANCE":"mean"

})

bureau_balance_agg.columns=["AVG_MONTH_BALANCE"]

bureau_balance_agg.reset_index(inplace=True)
```

```python
previous_application.head()
```

```python
previous_agg = previous_application.groupby("SK_ID_CURR").agg({

"AMT_APPLICATION":"mean",

"AMT_CREDIT":"mean",

"AMT_ANNUITY":"mean",

"CNT_PAYMENT":"mean",

"SK_ID_PREV":"count"

})

previous_agg.columns=[

"AVG_PREV_APPLICATION",

"AVG_PREV_CREDIT",

"AVG_PREV_ANNUITY",

"AVG_PREV_PAYMENT",

"TOTAL_PREVIOUS_LOAN"

]

previous_agg.reset_index(inplace=True)

previous_agg.head()
```

```python
pos_CASH_balance.head()
```

```python
pos_agg = pos_CASH_balance.groupby("SK_ID_CURR").agg({

"MONTHS_BALANCE":"mean",

"CNT_INSTALMENT":"mean",

"CNT_INSTALMENT_FUTURE":"mean"

})

pos_agg.columns=[

"AVG_POS_MONTH",

"AVG_INSTALLMENT",

"AVG_FUTURE_INSTALLMENT"

]

pos_agg.reset_index(inplace=True)
```

```python
credit_card_balance.head()
```

```python
credit_agg = credit_card_balance.groupby("SK_ID_CURR").agg({

"AMT_BALANCE":"mean",

"AMT_CREDIT_LIMIT_ACTUAL":"mean",

"AMT_PAYMENT_TOTAL_CURRENT":"mean"

})

credit_agg.columns=[

"AVG_CARD_BALANCE",

"AVG_CARD_LIMIT",

"AVG_CARD_PAYMENT"

]

credit_agg.reset_index(inplace= True)
```

```python
installments_payments.head()
```

```python
installment_agg = installments_payments.groupby("SK_ID_CURR").agg({

"AMT_INSTALMENT":"mean",

"AMT_PAYMENT":"mean",

"DAYS_ENTRY_PAYMENT":"mean"

})

installment_agg.columns=[

"AVG_INSTALLMENT_AMOUNT",

"AVG_PAYMENT",

"AVG_PAYMENT_DAY"

]

installment_agg.reset_index(inplace=True)
```

```python
final_df = application_train.merge(
bureau_agg,

on="SK_ID_CURR",

how="left"
)
```

```python
# merge bureau balance
final_df = final_df.merge(
bureau_balance_agg,

on="SK_ID_CURR",

how="left"
)
```

```python
# Merge previous application

final_df = application_train.merge(
previous_agg,

on="SK_ID_CURR",

how="left"
)
```

```python
# Merge POS

final_df =final_df .merge(
pos_agg,

on="SK_ID_CURR",

how="left"
)
```

```python
# MERGE credit card

final_df =final_df .merge(
credit_agg,

on="SK_ID_CURR",

how="left"
)
```

```python
# merge installments

final_df =final_df .merge(
installment_agg,

on="SK_ID_CURR",

how="left"
)
```

```python

# Machine Learning Libraries

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

from sklearn.impute import SimpleImputer

from sklearn.pipeline import Pipeline

from sklearn.compose import ColumnTransformer

# Models

from sklearn.linear_model import LogisticRegression

from sklearn.tree import DecisionTreeClassifier

from sklearn.ensemble import RandomForestClassifier

from sklearn.ensemble import ExtraTreesClassifier

from sklearn.ensemble import GradientBoostingClassifier

from sklearn.ensemble import AdaBoostClassifier

# Evaluation

from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_auc_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import RocCurveDisplay

import joblib

print("Machine Learning Libraries Imported")

```

```python
df = pd.read_csv("Final_Home_Loan_Default_Dataset.csv")
df.head()
```

```python
X = df.drop("TARGET", axis=1)

y = df["TARGET"]

print(X.shape)

print(y.shape)
```

```python
num_cols = X.select_dtypes(include=["int64","float64"]).columns

cat_cols = X.select_dtypes(include=["object"]).columns

print("Numerical :",len(num_cols))

print("Categorical :",len(cat_cols))
```

```python
label_encoder = LabelEncoder()

for col in cat_cols:
    X[col] = label_encoder.fit_transform(X[col].astype(str))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    
X,
y,
test_size=0.20,

random_state=42,

stratify = y

)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
print(X_train.shape)

print(X_test.shape)
```

```python
def evaluate_model(model):

    prediction = model.predict(X_test)

    probability = model.predict_proba(X_test)[:,1]

    accuracy = accuracy_score(y_test,prediction)

    precision = precision_score(y_test,prediction)

    recall = recall_score(y_test,prediction)

    f1 = f1_score(y_test,prediction)

    auc = roc_auc_score(y_test,probability)

    return accuracy,precision,recall,f1,auc
```

```python
# Logistic Regression

lr = LogisticRegression(max_iter=1000)

lr.fit(X_train,y_train)
```

```python
lr_result = evaluate_model(lr)

lr_result
```

```python
# Decision Tree

dt = DecisionTreeClassifier(
random_state=42
)

dt.fit(X_train,y_train)
```

```python
dt_result = evaluate_model(dt)

dt_result
```

```python
# Random Forest

rf = RandomForestClassifier(

    n_estimators=200,

    max_depth=10,

    random_state=42,

n_jobs=-1

)

rf.fit(X_train,y_train)
```

```python
rf_result = evaluate_model(rf)

rf_result
```

```python
# Extra Trees

et = ExtraTreesClassifier(

n_estimators=200,

random_state=42,
    
n_jobs=-1

)

et.fit(X_train,y_train)
    
```

```python
et_result = evaluate_model(et)

et_result
```

```python
gb = GradientBoostingClassifier(

random_state=42

)

gb.fit(X_train,y_train)

```

```python
gb_result=evaluate_model(gb)

gb_result
```

```python
# AdaBoost

ada = AdaBoostClassifier(

random_state=42

)

ada.fit(X_train,y_train)
```

```python
ada_result=evaluate_model(ada)

ada_result
```
